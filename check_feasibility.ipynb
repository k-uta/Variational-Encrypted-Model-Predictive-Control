```python
import numpy as np
from vempc.core.mpc import MPCProblem
from vempc.core.constraints import ConstraintPenalty

# Setup simple box constraints: -1 <= u <= 1
N, m = 10, 1
G = np.vstack([np.eye(N*m), -np.eye(N*m)])
h_func = lambda x0: np.ones(2*N*m)

penalty = ConstraintPenalty(G, h_func, mode='indicator')

# Test feasible sequence
U_feasible = np.ones(N*m) * 0.5
x0 = np.zeros(2)
print(f"Feasible weight: {penalty.desirability_weight(U_feasible, x0)}") # expected 1.0 since strictly within the constraints

# Test infeasible sequence
U_infeasible = np.ones(N*m) * 1.1
print(f"Infeasible weight: {penalty.desirability_weight(U_infeasible, x0)}") # expected 0.0 since strictly outside the constraints

```
