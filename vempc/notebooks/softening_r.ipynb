```python
import numpy as np
import matplotlib.pyplot as plt

gamma = np.linspace(-2.0, 2.0, 2000)

s = np.maximum(gamma, 0.0)

r_hard = (gamma <= 0).astype(float)
etas = [1, 2, 5, 10, 20]
soft_weights = {eta: np.exp(-eta * s) for eta in etas}

# Plot
plt.figure()
plt.plot(gamma, r_hard, label="r (indicator)")

for eta in etas:
    plt.plot(gamma, soft_weights[eta], label=f"exp(-eta s), eta={eta}")

plt.xlabel("Residual gamma")
plt.ylabel("Weight r")
plt.title("Hard Indicator vs Soft Exponential Relaxation")
plt.legend()
plt.show()
```
